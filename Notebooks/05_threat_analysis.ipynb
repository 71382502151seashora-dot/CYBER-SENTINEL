# 05 - Cybersecurity Threat Analysis
Objective: analyze existing labels if present; otherwise use transparent rule-based screening. A flag is not proof of malicious activity.

```python
import sys
sys.path.append('..')
from src.analysis_utils import load_data, prepare_data, create_rule_flags

df = load_data('../Dataset/cybersecurity_data.csv')
df, mapping = prepare_data(df)
if 'label' in mapping:
    labels = df[mapping['label']].fillna('Unknown').astype(str).str.strip()
    display(labels.value_counts().to_frame('records'))
    normal = labels.str.lower().str.contains('benign|normal', regex=True)
    print('Normal/Benign:', int(normal.sum()))
    print('Other labelled:', int((~normal & labels.str.lower().ne('unknown')).sum()))
else:
    flags, reasons, rules = create_rule_flags(df, mapping)
    print('No attack labels found. Rule thresholds:')
    print('\n'.join(rules) if rules else 'No usable numeric fields.')
    print('Records needing review:', int(flags.sum()))
```

```python
import pandas as pd
if 'label' in mapping and 'protocol' in mapping:
    from src.analysis_utils import protocol_series
    display(pd.crosstab(protocol_series(df, mapping), df[mapping['label']].astype(str)))
else:
    print('Label/protocol cross-analysis requires both fields.')
```
