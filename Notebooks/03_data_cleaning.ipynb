# 03 - Data Cleaning
Objective: standardize column names, parse known numeric/time fields, identify missing values and remove duplicate rows.

```python
import sys
sys.path.append('..')
from src.analysis_utils import load_data, prepare_data

df = load_data('../Dataset/cybersecurity_data.csv')
prepared, mapping = prepare_data(df)
print('Field mapping:', mapping)
print('Original shape:', prepared.shape)
print('Duplicate rows:', prepared.duplicated().sum())
```

```python
cleaned = prepared.drop_duplicates().copy()
print('Shape after duplicate removal:', cleaned.shape)
print('Missing values after cleaning:')
print(cleaned.isna().sum().sort_values(ascending=False).head(25))
cleaned.to_csv('../outputs/cleaned_data_from_notebook.csv', index=False)
print('Cleaned file saved in outputs/')
```
