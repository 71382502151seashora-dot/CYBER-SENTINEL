# 04 - Traffic Analysis
Objective: summarize traffic by protocol, source IP, destination port, packet count, byte count, and timestamp when available.

```python
import pandas as pd
import sys
sys.path.append('..')
from src.analysis_utils import load_data, prepare_data, protocol_series

df = load_data('../Dataset/cybersecurity_data.csv')
df, mapping = prepare_data(df)
print('Detected fields:', mapping)
protocols = protocol_series(df, mapping)
if protocols is not None:
    display(protocols.value_counts().to_frame('records'))
else:
    print('Protocol field is unavailable.')
```

```python
if 'source_ip' in mapping:
    display(df[mapping['source_ip']].astype(str).value_counts().head(10).to_frame('records'))
if 'destination_port' in mapping:
    display(df[mapping['destination_port']].value_counts().head(10).to_frame('records'))
for key in ['packet_count', 'flow_duration', 'byte_count']:
    if key in mapping:
        print('\n', key, '\n', df[mapping[key]].describe())
```

```python
if 'timestamp' in mapping:
    times = pd.to_datetime(df[mapping['timestamp']], errors='coerce')
    print('Valid timestamps:', times.notna().sum())
    display(times.dt.floor('h').value_counts().sort_index().head(20))
else:
    print('Timestamp field is unavailable.')
```
